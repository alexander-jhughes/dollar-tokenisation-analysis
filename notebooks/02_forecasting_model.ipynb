```python
import torch
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
# Load your stablecoin data
df = pd.read_csv('/Users/alexhughes/Desktop/Project Git DLT/Project 2/petrodollar-rails/data/raw/stablecoin_supply.csv')
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').reset_index(drop=True)

# Extract just the supply column
supply = df['usd_stablecoin_supply'].values

# Split 80/20
split_point = int(len(supply) * 0.8)
train_supply = supply[:split_point]
test_supply = supply[split_point:]

print(f"Train: {len(train_supply)} points, Test: {len(test_supply)} points")
```

```python
# Normalise to 0-1
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_supply.reshape(-1, 1)).flatten()
test_scaled = scaler.transform(test_supply.reshape(-1, 1)).flatten()

# Create windowed data (30-day windows)
def create_windows(data, window_size=30):
    X, y = [], []
    for i in range(len(data) - window_size):
        X.append(data[i:i+window_size])
        y.append(data[i+window_size])
    return np.array(X), np.array(y)

X_train, y_train = create_windows(train_scaled, window_size=30)
X_test, y_test = create_windows(test_scaled, window_size=30)

print(f"Training windows: {X_train.shape}")
print(f"Test windows: {X_test.shape}")
```
