learn anything a classical model in `02_forecasting_model.ipynb` can't. Using
TensorFlow there would be the wrong tool for the data, not a skill flex.

which is high-frequency enough (potentially thousands of points if pulled from
historical block data) to justify the comparison. The point of this notebook is
not "use TensorFlow" - it's "honestly test whether TensorFlow earns its place
here", and report the answer either way.

1. Pull historical on-chain confirmation time / gas fee series
2. Build the same classical baseline approach as in notebook 02 (train/test split, naive baseline)
3. Build a small LSTM or dense net (Keras/TensorFlow) on the same train/test split
4. Compare on the same error metric, no cherry-picking
